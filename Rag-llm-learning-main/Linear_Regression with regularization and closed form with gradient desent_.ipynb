```python
x=np.array([1,2,3,4,5])
y=np.array([3,5,7,9,11])
```

```python
#formula y=wx+b
w=0
b=0
y_pred=w*x + b
```

```python
m=len(y)
cost = (1 / m) * np.sum((y - y_pred) ** 2)
print("Predictions:",y_pred)
print("cost",cost)
```

```python
#gradients
lr=0.001
epochs=1000
cost_hist=[]
for i in range(epochs):
  y_pred=w*x+b
  dw = (-2 / m) * np.sum(x * (y - y_pred))
  db = (-2 / m) * np.sum(y - y_pred)
  w = w - lr * dw
  b = b - lr * db
  cost = (1 / m) * np.sum((y - y_pred) ** 2)
  cost_hist.append(cost)
print("Final w:", w)
print("Final b:", b)
print("Final cost:", cost_hist[-1])
```

```python
#closed form
import numpy as np
import matplotlib.pyplot as plt

X = np.array([1, 2, 3, 4, 5])
y = np.array([3, 5, 7, 9, 11])

X_b = np.c_[np.ones(len(X)), X]

theta = np.linalg.inv(X_b.T @ X_b) @ X_b.T @ y

b = theta[0]
w = theta[1]

y_pred = X_b @ theta

m = len(y)
cost = (1 / m) * np.sum((y - y_pred) ** 2)

print("b:", b)
print("w:", w)
print("Predictions:", y_pred)
print("Cost:", cost)

plt.scatter(X, y)
plt.plot(X, y_pred)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

X = np.array([1, 2, 3, 4, 5])
y = np.array([3, 5, 7, 9, 11])

w = 0
b = 0

learning_rate = 0.01
lambda_ = 0.1
epochs = 1000

m = len(y)

cost_history = []

for i in range(epochs):

    y_pred = w * X + b

    dw = (-2 / m) * np.sum(X * (y - y_pred)) + 2 * lambda_ * w
    db = (-2 / m) * np.sum(y - y_pred)

    w = w - learning_rate * dw
    b = b - learning_rate * db

    cost = (1 / m) * np.sum((y - y_pred) ** 2) + lambda_ * w**2
    cost_history.append(cost)

y_pred = w * X + b

print("w:", w)
print("b:", b)
print("Predictions:", y_pred)
print("Final Cost:", cost_history[-1])

plt.plot(cost_history)
plt.xlabel("Epochs")
plt.ylabel("Cost")
plt.title("Ridge Regression Cost")
plt.show()
```
