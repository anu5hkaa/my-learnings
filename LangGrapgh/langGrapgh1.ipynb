```python
from typing import TypedDict

from langgraph.graph import StateGraph, START, END
```

```python
class StudentState(TypedDict):
  name:str
  marks:int
  status:str
  grade:str
  result:str
```

```python
from ast import Return
def check_marks(state:StudentState):
  if state["marks"]>=40:
    state["status"]='Pass'
  else:
     state['status']='Fail'
  return state
```

```python
def check_grade(state:StudentState):
  if state['status']=="Fail":
    state['grade']=='F'
  elif state['marks']>=90:
    state['grade']="A"
  elif state['marks']>=75:
    state['grade']="B"
  elif state['marks']>=60:
    state['grade']="C"
  else:
    state['grade']="D"
  return state
```

```python
def create_results(state:StudentState):
  state['result']=(
      f"{state['name']} scored {state['marks']} marks"
      f"and recieved {state['grade']}."
  )
  return state
```

```python
graph=StateGraph(StudentState)
graph.add_node("check_marks", check_marks)
graph.add_node("calculate_grade", check_grade)
graph.add_node("create_result", create_results)
```

```python
graph.add_edge(START, "check_marks")
graph.add_edge("check_marks", "calculate_grade")
graph.add_edge("calculate_grade", "create_result")
graph.add_edge("create_result", END)
```

```python
app = graph.compile()
input_data = {
    "name": "Anushka",
    "marks": 75,
    "status": "",
    "grade": "",
    "result": ""
}


# 10. Run the graph
result = app.invoke(input_data)


# 11. Print the result
print(result)
print(result["result"])
```
